A tephigram of the whole sounding with a light, sparse background (isobars,
isotherms and dry adiabats only, without labels) so that it stays readable at
thumbnail size.

```python
grey = {"color": "#c8c8c8", "linewidth": 0.5, "alpha": 1, "labels": False}
sparse = {**grey, "interval": 20.0}

tephi = ekp.Tephigram(figsize=(2, 2), grid=False)
tephi.grid(
    isobars={**grey, "values": [1000, 850, 700, 500, 300, 200, 100]},
    isotherms=sparse,
    dry_adiabats=sparse,
    saturated_adiabats=False,
    mixing_ratios=False,
)

tephi.temperature(temperature, p=pressure, color="#F94144", linewidth=1.5)
tephi.dewpoint(dewpoint, p=pressure, color="#277DA1", linestyle="-", linewidth=1.5)

tephi.ax.set_axis_off()

tephi.save(THUMBNAILS_DIR / "thermodynamic-tephigram.png", transparent=True)
```

A skew-T with the same light background and colours.
